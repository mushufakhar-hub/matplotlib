```python
food= np.array(["Apple","bananan","kiwi","apricots","mango"])
amount=np.array([12,14,8,16,20])
plt.bar(food,amount)
plt.title("Fruits Eat in Months",fontsize=25,
          c="g")
plt.xlabel("\n-------<<<<<<<<---Fruits----->>>>>>>>------")
plt.ylabel("\n--------<<<<<<<<------Quantity----->>>>>>>-------")



plt.show()
```

```python
food= ["Apple","bananan","kiwi","apricots","mango"]
amount=np.array([12,14,8,16,20])
plt.pie(amount, labels=food,autopct="%1.1f%%", explode=[0,0.05,0,0.09,0],
        shadow=True
        )
plt.title("Fruits Eat in Months",fontsize=25,
            c="g")
plt.legend(title="Fruits", loc="upper left", bbox_to_anchor=(1.1, 0.5))
# plt.xlabel("\n-------<<<<<<<<---Fruits----->>>>>>>>------")
# plt.ylabel("\n--------<<<<<<<<------Quantity----->>>>>>>-------")



plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
study1 = np.array([1,2,3,4,5,6])
score1 = np.array([50,65,69,76,79,72])

study2 = np.array([2,3,5,6,7])
score2 = np.array([55,63,75,70,80])
plt.scatter(study1 , score1,alpha=0.5,c="g",s=100,label="student1")



plt.scatter(study2 , score2,c="r",s=120,label="student2")
plt.title("Graph of Student marks", fontsize=28)
plt.xlabel("\n------------Hours---------",fontsize=18)
plt.ylabel("--------Marks-------",fontsize=18)
# plt.grid(axis="y",
#          lw=2 ,
#          ls="--")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
z= np.array([3,5,9])
x ,y=plt.subplots(2,2)
y[0,0].plot(z , z*2 ,c="g")
y[0,0].set_title("multiply by 2")

y[0,1].plot(z , z**2 ,c="r")
y[0,1].set_title("square")

y[1,0].plot(z , z**3 ,c="b")
y[1,1].set_title("square")

y[1,1].plot(z , z**2 ,c="y")
y[1,1].set_title("square")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv("data.csv")
# print(df)
total_count=(df["Type1"].value_counts())
plt.bar(total_count.index,total_count.values,color="skyblue",edgecolor="g")
plt.title("Data FROM CSV File",fontsize=30,c="y")
plt.xlabel("\n----------Type--------")
plt.ylabel("----------Count--------")

plt.show()
```
